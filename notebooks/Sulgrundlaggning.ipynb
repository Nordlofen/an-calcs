```python
from an_calcs.notebook import Grundplan

plan = Grundplan(titel="Sulgrundläggning")
plan
```

Välj **Öppna ritning**, sedan **+ Väggsula** eller **+ Pelarsula** och klicka på ritningen. Ändra littera och indata, tryck **Beräkna**, sedan **Minimera**. Klicka på taggen för att redigera igen.

Dra en etikett för att flytta den. Välj **Kopiera sula** i dialogen och klicka på ritningen för att placera en ny sula med samma indata. Kopian får eget littera; anpassa last och geometri och beräkna på nytt.

Reglaget **Etikettstorlek** styr etiketternas grundstorlek. Etiketterna förstoras och förminskas tillsammans med ritningen när du zoomar. Grundstorleken och etiketternas lägen sparas i projektet.

Väggsulans laster anges **per meter vägg**, pelarsulans laster som **totallaster**. Startvärdena är exempel och ska anpassas. Mått och laster läses inte automatiskt från ritningen.

Ange brottlasterna under **Laster – Brott**. För isolerad sula: kryssa i **Underliggande isolering** under **Isolering**, ange färdiga dimensionerande bärförmågor `f_d.brott` och `f_d.bruk` i kPa samt långtidslasterna under **Laster – Bruk**. Egentyngd tillkommer med faktor 1,5 i brott och 1,0 i bruk.

Ange moment direkt vid sulan, både i brott och bruk. Inget momentbidrag från horisontallast gånger hävarm läggs till. Horisontallaster i brott påverkar fortfarande jordens bärighet.

Isoleringskontrollen använder `N / (b_eff × l_eff)` med separata effektiva areor för brott och bruk. Isoleringen förutsätts täcka hela arean. Resultatet visar jord i brott, isolering i brott och isolering i bruk; etiketten visar den högsta utnyttjandegraden och vilken kontroll som styr. Krypdeformation och sättning beräknas inte.
